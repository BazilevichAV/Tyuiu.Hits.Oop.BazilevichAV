```
using System;

public interface IPerson
{
    string Name { get; set; }
    int Age { get; set; }
    string Gender { get; set; }

    string GetInfo();
    void SayHello();
}

public class Person : IPerson
{
    private string _name;
    private int _age;
    private string _gender;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public static int TotalPeople { get; private set; } = 0;

    public Person()
    {
        Name = "Безымянный";
        Age = 0;
        Gender = "Не указан";
        TotalPeople++;
    }

    public Person(string name, int age, string gender) : this()
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }

    public static void DisplayTotalPeople()
    {
        Console.WriteLine($"Общее количество людей: {TotalPeople}");
    }

    protected void SetAge(int value)
    {
        Age = value;
    }

    public void Interact(Person other)
    {
        Console.WriteLine($"{Name} общается с {other.Name}");
    }

    public void IntroduceTo(Person other)
    {
        Interact(other);
        SayHello();
        other.SayHello();
    }
}

public class Student : Person
{
    private string _university;
    private int _course;

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public int Course
    {
        get { return _course; }
        set
        {
            if (value >= 0)
                _course = value;
            else
                throw new ArgumentOutOfRangeException("Курс не может быть отрицательным!");
        }
    }

    public Student(string name, int age, string gender, string university, int course)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    public Student(string name, int age, string gender, string university, int course, int newAge)
        : this(name, age, gender, university, course)
    {
        SetAge(newAge);
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я студент {Course}-го курса в {University}.");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} учится в {University} на {Course}-м курсе.");
    }
}

public class Employee : Person
{
    private string _company;
    private decimal _salary;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public decimal Salary
    {
        get { return _salary; }
        set
        {
            if (value >= 0)
                _salary = value;
            else
                throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной!");
        }
    }

    public Employee(string name, int age, string gender, string company, decimal salary)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    public Employee(string name, int age, string gender, string company, decimal salary, decimal newSalary)
        : this(name, age, gender, company, salary)
    {
        Salary = newSalary;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Зарплата: {Salary} руб.";
    }

    public void GiveRaise(decimal amount)
    {
        Salary += amount;
        Console.WriteLine($"{Name} получил повышение. Новая зарплата: {Salary} руб.");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в компании {Company}.");
    }
}

public class Teacher : Person
{
    private string _subject;
    private int _experience;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public int Experience
    {
        get { return _experience; }
        set
        {
            if (value >= 0)
                _experience = value;
            else
                throw new ArgumentOutOfRangeException("Стаж не может быть отрицательным!");
        }
    }

    public Teacher(string name, int age, string gender, string subject, int experience)
        : base(name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    public Teacher(string name, int age, string gender, string subject, int experience, int newExperience)
        : this(name, age, gender, subject, experience)
    {
        Experience = newExperience;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я преподаю {Subject}.");
    }

    public void Teach(Student student)
    {
        Console.WriteLine($"{Name} преподаёт {Subject} студенту {student.Name}");
    }

    public void GiveLecture(Employee employee)
    {
        Console.WriteLine($"{Name} читает лекцию по предмету {Subject} сотруднику {employee.Name} в компании {employee.Company}");
    }
}

var student = new Student("Алена", 20, "Женский", "ТИУ", 3);
var employee = new Employee("Иван", 30, "Мужской", "ООО Безопасность", 50000m);
var teacher = new Teacher("Юлия", 45, "Женский", "Математика", 20);

Console.WriteLine("Студент:");
Console.WriteLine(student.GetInfo());
student.SayHello();
student.Study();
Console.WriteLine();

Console.WriteLine("Работник:");
Console.WriteLine(employee.GetInfo());
employee.SayHello();
employee.GiveRaise(10000m);
Console.WriteLine();

Console.WriteLine("Преподаватель:");
Console.WriteLine(teacher.GetInfo());
teacher.SayHello();

Console.WriteLine();

student.IntroduceTo(employee);

Console.WriteLine();

teacher.Teach(student);

Console.WriteLine();

teacher.GiveLecture(employee);

Console.WriteLine();

Person.DisplayTotalPeople();
```
