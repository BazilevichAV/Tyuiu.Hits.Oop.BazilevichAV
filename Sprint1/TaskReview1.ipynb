```
using System;
using System.Collections.Generic;

public class Product
{  // Модификаторы доступа
    private string name; // только внутри класса
    private decimal price;
    private int quantity;
    protected string category;  // в классе и наследниках
    internal string description; // в пределах сборки

    public string Name  // геттеры и сеттеры
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {  // создаем через свойства
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }
     // 3 метода
    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity = amount;
        Console.WriteLine($"Количество товара \"{name}\" обновлено: {quantity} шт.");
    }

    public string GetProductInfo()
    {
        return $"Товар: {name}, Цена: {price} руб., Количество: {quantity} шт., Категория: {category}, Описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products;  // храним список товаров продавца

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
        products = new List<Product>();
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Продавец {name} добавил товар: {product.Name}");
    }

    public void SellProduct(Product product, int quantity)
    {
        Console.WriteLine($"Продавец {name} продал {quantity} шт. товара \"{product.Name}\"");
        product.UpdateQuantity(product.Quantity - quantity);
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name}, ID: {employeeId}, Зарплата: {salary} руб., Контакты: {contactInfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers;  // взаимодеййствие объектов
    private List<Product> products;

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
        sellers = new List<Seller>();
        products = new List<Product>();
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин \"{storeName}\"");
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине \"{storeName}\":");
        foreach (var p in products)
        {
            Console.WriteLine($"{p.GetProductInfo()}");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, Адрес: {location}, Часы работы: {storeHours}";
    }
}

var product1 = new Product("Хлеб", 50m, 100, "Продукты", "Свежий белый хлеб");
Console.WriteLine("Товар");
Console.WriteLine(product1.GetProductInfo());
Console.WriteLine($"Общая стоимость: {product1.GetTotalPrice()} руб.");
product1.UpdateQuantity(80);
Console.WriteLine();

var seller = new Seller("Иван", "A001", 20000m, "ivan@mail.ru");
Console.WriteLine("Продавец");
Console.WriteLine(seller.GetSellerInfo());
seller.AddProduct(product1);
seller.SellProduct(product1, 30);
Console.WriteLine();

var store = new Store("Пятёрочка", "г. Москва, ул. Ленина, д. 1", "08:00 - 22:00");
Console.WriteLine("Магазин");
Console.WriteLine(store.GetStoreInfo());
store.AddSeller(seller);
store.AddProduct(product1);
store.ListProducts();
```
