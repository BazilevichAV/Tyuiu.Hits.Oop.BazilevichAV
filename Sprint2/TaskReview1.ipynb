```
using System;
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLead
{
    string[] GetTeamMembers();
}

// Базовый
public class Employee
{
    private string _name;
    private string _position;
    private double _salary;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public double Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }
    
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}, Должность: {Position}, Зарплата: {Salary} руб.");
    }
}

// Простоое Manager
public class Manager : Employee
{
    private string _department;

    public string Department
    {
        get { return _department; }
        set { _department = value; }
    }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

// Сложное + Множественное
public class ProjectManager : Manager, IReportable, ITeamLead
{
    private string _projectName;
    private string[] _teamMembers;

    public string ProjectName
    {
        get { return _projectName; }
        set { _projectName = value; }
    }

    public ProjectManager(string name, string position, double salary, string department,
                          string projectName, string[] teamMembers)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        _teamMembers = teamMembers;
    }
    // Название проекта
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    // IReportable
    public string GenerateReport()
    {
    return $"Отчёт по проекту \"{ProjectName}\": отдел {Department}, команда из {_teamMembers.Length} человек.";
    
    }
    // ITeamLead
    public string[] GetTeamMembers()
    {
        return _teamMembers;
    }
}
Employee employee = new Employee("Иван", "Инженер", 50000);
Manager manager = new Manager("Сергей", "Менеджер", 80000, "Разработка");
ProjectManager projectManager = new ProjectManager("Юлия", "Проектный менеджер", 100000, "Разработка", "CRM система", new string[] { "Иван", "Сергей", "Юлия" });

Console.WriteLine("Employee (базовый)");
employee.PrintInfo();
Console.WriteLine();

Console.WriteLine("Manager (простое)");
manager.PrintInfo();
Console.WriteLine();

Console.WriteLine("ProjectManager (сложное)");
projectManager.PrintInfo();
Console.WriteLine();

Console.WriteLine("Отчёт (IReportable + GenerateReport)");
Console.WriteLine(projectManager.GenerateReport());
Console.WriteLine();

Console.WriteLine("Команда проекта (ITeamLead + GetTeamMembers)");
foreach (var member in projectManager.GetTeamMembers())
Console.WriteLine($"{member}");
```
