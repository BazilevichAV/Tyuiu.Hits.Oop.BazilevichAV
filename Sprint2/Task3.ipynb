### Вариант задания 1

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Person в C#, который будет представлять информацию о человеке. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Person:

• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).

• Методы:

o GetInfo(): метод для получения информации о человеке в виде строки.

o SayHello(): метод для вывода приветствия от человека.

Требования к производным классам:

Студент (Student): Должен содержать дополнительные атрибуты, такие как Университет (University) и Курс (Course). Метод SayHello() должен быть переопределен для добавления информации о курсе при приветствии.

Работник (Employee): Должен содержать дополнительные атрибуты, такие как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть переопределен для включения информации о компании и зарплате.

Преподаватель (Teacher) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы (Experience). Метод SayHello() должен быть переопределен для добавления информации о преподаваемом предмете при приветствии.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
public interface IWorker
{
    void Work();
}
// Базовый
public class Person
{
    private string _name;
    private int _age;
    private string _gender;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public static int TotalPeople { get; private set; } = 0;

    public Person()
    {
        Name = "Безымянный";
        Age = 0;
        Gender = "Не указан";
        TotalPeople++;
    }

    public Person(string name, int age, string gender) : this()
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }

    public void CelebrateBirthday()
    {
        Age++;
        Console.WriteLine($"{Name} отмечает день рождения. Новый возраст: {Age} лет.");
    }
}

// Простое
public class Student : Person
{
    private string _university;
    private int _course;
    private string _faculty;

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public int Course
    {
        get { return _course; }
        set
        {
            if (value >= 0)
                _course = value;
            else
                throw new ArgumentOutOfRangeException("Курс не может быть отрицательным!");
        }
    }

    public string Faculty
    {
        get { return _faculty; }
        set { _faculty = value; }
    }

    public Student(string name, int age, string gender, string university, int course, string faculty)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
        Faculty = faculty;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я студент {Course}-го курса в {University}.");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} учится в {University} на {Course}-м курсе.");
    }

    public void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен на факультете {Faculty}.");
    }
}

// Множественное
public class Employee : Person, IWorker
{
    private string _company;
    private decimal _salary;
    private string _position;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public decimal Salary
    {
        get { return _salary; }
        set
        {
            if (value >= 0)
                _salary = value;
            else
                throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной!");
        }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public Employee(string name, int age, string gender, string company, decimal salary, string position)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
        Position = position;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Зарплата: {Salary} руб., Должность: {Position}";
    }

    public void GiveRaise(decimal amount)
    {
        Salary += amount;
        Console.WriteLine($"{Name} получил повышение. Новая зарплата: {Salary} руб.");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в компании {Company} на должности {Position}.");
    }
}

// Сложное
public class Teacher : Employee
{
    private string _subject;
    private int _experience;
    private string _degree;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public int Experience
    {
        get { return _experience; }
        set
        {
            if (value >= 0)
                _experience = value;
            else
                throw new ArgumentOutOfRangeException("Стаж не может быть отрицательным!");
        }
    }

    public string Degree
    {
        get { return _degree; }
        set { _degree = value; }
    }

    public Teacher(string name, int age, string gender, string company, decimal salary, string position,
                   string subject, int experience, string degree)
        : base(name, age, gender, company, salary, position)
    {
        Subject = subject;
        Experience = experience;
        Degree = degree;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я преподаю {Subject}.");
    }

    public void Teach(Student student)
    {
        Console.WriteLine($"{Name} преподаёт {Subject} студенту {student.Name}");
    }

    public void GradeExam()
    {
        Console.WriteLine($"{Name} проверяет экзаменационные работы по предмету {Subject}.");
    }
}
Student student = new Student("Алена", 20, "Женский", "ТИУ", 3, "Информатика");
Console.WriteLine("Простое: Person → Student");
Console.WriteLine(student.GetInfo());
student.SayHello();
student.Study();
student.TakeExam();
student.CelebrateBirthday();
Console.WriteLine();

Teacher teacher = new Teacher("Юлия", 45, "Женский", "ТИУ", 60000m, "Доцент", "Математика", 20, "Кандидат наук");
Console.WriteLine("Сложное: Person → Employee → Teacher");
Console.WriteLine(teacher.GetInfo());
teacher.SayHello();
teacher.Work();
teacher.GiveRaise(5000m);
teacher.Teach(student);
teacher.GradeExam();
Console.WriteLine();

Employee employee = new Employee("Иван", 30, "Мужской", "ООО Безопасность", 50000m, "Инженер");
Console.WriteLine("Множественное: Person + IWorker → Employee");
Console.WriteLine(employee.GetInfo());
employee.SayHello();
employee.Work();
employee.GiveRaise(10000m);
Console.WriteLine();

Console.WriteLine($"Общее количество людей: {Person.TotalPeople}");
```
