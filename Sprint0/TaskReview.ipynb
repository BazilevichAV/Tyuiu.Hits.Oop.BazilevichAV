```
public interface IPlant // Класс Растение
{
    string Name { get; set; } // Имя
    string Type { get; set; } // Тип
    float Height { get; set; } // Высота

    string Bloom();  // Метод цветения
    void Grow(float amount); // Метод растет
}
public interface IFlower : IPlant // Класс Цветок
{
    string Color { get; set; }  // Цвет
    int NumberOfPetals { get; set; }  // Количество
}
public interface ITree : IPlant // Класс Дерево
{
    int Age { get; set; }  // Возраст
    string LeafType { get; set; } // ТипЛистьев
}

public class Plant : IPlant  // Реализует интерфейс Растения
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual string Bloom()
    {
        return$"Растение {Name} расцветает.";
    }

    public virtual void Grow(float amount)
    {
        Height += amount;
        Console.WriteLine($"Растение {Name} выросло на {amount} см. Текущая высота: {Height} см.");
    }
}

public class Flower : Plant, IFlower  // Реализует интерфейс Цветок
{
    public string Color { get; set; }
    public int NumberOfPetals { get; set; }

    public Flower(string name, string type, float height, string color, int numberOfPetals)
        : base(name, type, height)
    {
        Color = color;
        NumberOfPetals = numberOfPetals;
    }

    public override string Bloom()
    {
        return $"Цветок {Name} расцветает и радует глаз.";
    }
    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant, ITree // Реализует интерфейс Дерево
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, string type, float height, int age, string leafType)
        : base(name, type, height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override string Bloom()
    {
        return $"Дерево {Name} покрывается новыми листьями.";
    }
    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} растет. Новый возраст: {Age} лет.");
    }
}

// Тестирование
var flower = new Flower("Роза", "Цветок", 15.5f, "Красный", 20);
var tree = new Tree("Дуб", "Дерево", 200.0f, 5, "Лиственные");

// Вывод информации
Console.WriteLine("Цветок:");
Console.WriteLine(flower.Bloom());
flower.Grow(2.5f);
flower.Wither();
Console.WriteLine();

Console.WriteLine("Дерево:");
Console.WriteLine(tree.Bloom());
tree.Grow(10.0f);
tree.GrowOlder(1);
```
